```python
def md5_file(path: str | Path, chunk_size: int = 8192) -> str:
    """
    Calcula el MD5 de un archivo leyendo en bloques.
    """
    md5 = hashlib.md5()
    path = Path(path)

    with path.open("rb") as f:
        for chunk in iter(lambda: f.read(chunk_size), b""):
            md5.update(chunk)
    
    return md5.hexdigest()
```
